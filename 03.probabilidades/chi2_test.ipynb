# Chi-Cuadrado entre 2 variables nominales como outlook y play

Para realizar una prueba de chi-cuadrado (χ²) en Python utilizando el conjunto de datos `weather.nominal.csv` (conocido comúnmente como el set de datos "weather" de Weka con variables nominales como _outlook, temperature, humidity, windy y play_), debes cargar el archivo CSV con [pandas](https://pandas.pydata.org/), construir una tabla de contingencia con `pd.crosstab` y aplicar la función `chi2_contingency` de [SciPy](https://scipy.org/). [1, 2, 3, 4] 

## Código en Python para la prueba de independencia Chi-Cuadrado

```python
import pandas as pd
from scipy.stats import chi2_contingency

# 1. Cargar el dataset nominal del clima
# Asegúrate de que 'weather.nominal.csv' esté en tu directorio de trabajo
df = pd.read_csv('weather.nominal.csv')

# Visualizar las primeras filas para verificar los nombres de las columnas
print(df)

# 2. Crear una tabla de contingencia entre dos variables nominales (ej. 'outlook' y 'play')
# Puedes cambiar 'outlook' y 'play' por cualquier otra combinación categórica del CSV
tabla_contingencia = pd.crosstab(df['outlook'], df['play'])

print("\nTabla de Contingencia:")
print(tabla_contingencia)

# 3. Aplicar la prueba de Chi-cuadrado de independencia
chi2, p_valor, grados_libertad, valores_esperados = chi2_contingency(tabla_contingencia)

# 4. Mostrar los resultados estadísticos
print(f"\nEstadístico Chi2: {chi2:.4f}")
print(f"P-valor: {p_valor:.4f}")
print(f"Grados de libertad (df): {grados_libertad}")
print("\nValores esperados:")
print(valores_esperados)

# 5. Interpretación básica basada en un nivel de significancia alpha = 0.05
alpha = 0.05
if p_valor < alpha:
    print("\nResultado: Se rechaza la hipótesis nula (H0). Las variables son dependientes (existe asociación significativa).")
else:
    print("\nResultado: No se rechaza la hipótesis nula (H0). Las variables son independientes.")
```

## Explicación de los Componentes


* Carga de datos (pd.read_csv): Lee el archivo delimitado por comas transformándolo en un DataFrame de Pandas. [2] 
* Tabla de contingencia (pd.crosstab): Agrupa las frecuencias observadas cruzando las categorías de ambas columnas nominales (por ejemplo, cuántas veces estuvo sunny y se decidió yes o no jugar). [2] 
* Prueba estadística (chi2_contingency): Compara las frecuencias observadas contra las frecuencias teóricas esperadas bajo el supuesto de independencia. [2, 5] 
* P-valor (p_valor): Si el valor es inferior a 0.05, indica que la relación entre ambas condiciones climáticas y la decisión de jugar es estadísticamente significativa. [2, 6]

# Chi-Cuadrado con definción de variable objetivo y variables predictoras

Para evaluar la relación de todas las variables predictoras (`outlook`, `temperature`, `humidity`, `windy`) respecto a la variable objetivo (`play`), debes calcular una prueba de Chi-cuadrado de independencia para cada par.

El siguiente código en Python automatizado utilizando un ciclo `for` para evaluar todas las combinaciones posibles de manera ordenada:

```python
import pandas as pd
from scipy.stats import chi2_contingency

# 1. Cargar el dataset
df = pd.read_csv('weather.nominal.csv')

# 2. Definir la variable objetivo y las variables predictoras
objetivo = 'play'
predictores = ['outlook', 'temperature', 'humidity', 'windy']

print("--- RESULTADOS DE LA PRUEBA DE CHI-CUADRADO RESPECTO A 'PLAY' ---\n")

alpha = 0.05

# 3. Iterar sobre cada variable predictora para realizar la prueba
for var in predictores:
    # Crear la tabla de contingencia
    tabla = pd.crosstab(df[var], df[objetivo])
    
    # Aplicar la prueba de Chi-cuadrado
    chi2, p_valor, dof, esperados = chi2_contingency(tabla)
    
    # Determinar si existe relación estadísticamente significativa (alpha = 0.05)
    significativo = "SÍ" if p_valor < alpha else "NO"
    
    # Mostrar resultados limpios y formateados
    print(f"Variable: '{var}' vs '{objetivo}'")
    print(f"  - Estadístico Chi2: {chi2:.4f}")
    print(f"  - P-valor: {p_valor:.4f}")
    print(f"  - Grados de libertad: {dof}")
    print(f"  - ¿Es estadísticamente significativa? {significativo}")
    print(f"  - Tabla de contingencia:\n{tabla}")
    print(f"  - Frecuencias esperadas:\n{esperados}")
    print("-" * 50)
```

## 💡 Nota importante sobre este dataset

El dataset clásico `weather.nominal` tiene una muestra muy pequeña (**14 filas**). Debido a esto, la prueba de Chi-cuadrado podría arrojar p-valores altos (mayores a 0.05) indicando que "no hay significancia estadística", y es probable que recibas advertencias en la consola porque muchas frecuencias esperadas serán menores a 5. ## Implementación del Modelo Probabilístico (Estilo Naive Bayes)

### División del Dataset en Entrenamiento y Prueba

Para implementar un modelo probabilístico utilizando variables categóricas (nominales), el algoritmo ideal es Naive Bayes Categórico (`CategoricalNB`) de scikit-learn. Dado que los modelos matemáticos requieren valores numéricos, primero codificaremos tus variables de texto a números utilizando OrdinalEncoder y luego realizaremos la división del dataset.

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder
from sklearn.naive_bayes import CategoricalNB
from sklearn.metrics import accuracy_score, classification_report

# 1. Separar las variables predictoras (X) de la variable objetivo (y)
X = df[['outlook', 'temperature', 'humidity', 'windy']]
y = df['play']

# 2. Transformar las variables nominales (texto) a números para el modelo
# Naive Bayes Categórico requiere que las categorías estén codificadas numéricamente
encoder = OrdinalEncoder()
X_encoded = encoder.fit_transform(X)

# 4. División del dataset: 80% Entrenamiento y 20% Prueba
# Usamos 'stratify=y' para mantener la misma proporción de 'yes' y 'no' en ambos conjuntos
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, 
    test_size=0.20, 
    random_state=42, 
    stratify=y
)

print(f"Dimensiones del dataset original: {df.shape[0]} filas")
print(f"Filas asignadas a Entrenamiento (80%): {X_train.shape[0]}")
print(f"Filas asignadas a Prueba (20%): {X_test.shape[0]}\n")

# 4. Inicializar e implementar el Modelo Probabilístico (Naive Bayes Categórico)
modelo_probabilistico = CategoricalNB()
modelo_probabilistico.fit(X_train, y_train)

# 5. Realizar predicciones en el conjunto de prueba
y_pred = modelo_probabilistico.predict(X_test)

# 6. Evaluar el rendimiento del modelo
exactitud = accuracy_score(y_test, y_pred)
print(f"Precisión del modelo (Accuracy): {exactitud:.2%}\n")
print("Reporte de Clasificación detallado:")
print(classification_report(y_test, y_pred))
```
